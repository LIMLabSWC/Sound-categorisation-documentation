```python
import os
import glob
import pickle
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
from scipy.spatial.distance import cosine
from scipy.stats import spearmanr, ranksums

# =====================================================
# Settings
# =====================================================
root_dir = r"D:\sc_modelling-amir-2025"

species_list = ["Mouse", "Human"]

participants_by_species = {
    "Mouse": [
        "QP0100", "QP0101", "QP0103", "QP0121",
        "QP062", "QP063", "QP070", "QP071"
    ],
    "Human": [
        '6363339', '6363334', '6363353', '6363338', '6363346', '6363343',
        '6363329', '6363354', '6363347', '6363328', '6363330', '6363341',
        '6363407', '6363430', '6363368', '6363349', '6363364', '6269773',
        '6269770', '6269757', '6269881', '6357507', '6357495', '6357501',
        '6357497', '6357556', '6357698', '6357829'
    ]
}

samplers = ["CMA-ES", "TPE"]

folders = {
    "BE": "Hybrid_BE_fit",
    "SC": "Hybrid_SC_fit",
    "Hybrid": "Hybrid_new3_fit",
}

save_dir = os.path.join(root_dir, "cosine_similarity_parameters_BE_SC_Hybrid_range_normalized")
os.makedirs(save_dir, exist_ok=True)

# =====================================================
# Parameter search ranges
# =====================================================
param_ranges = {
    "sigma_noise":    (0.01, 0.45),
    "A_repulsion":    (0.0, 0.5),
    "gamma":          (0.01, 1.0),
    "sigma_update":   (0.01, 0.7),
    "eta_learning":   (0.01, 1.0),
    "sigma_boundary": (0.01, 10.0),
    "alpha":          (0.0, 1.0),
}

# =====================================================
# Vector definitions
# =====================================================
parameter_sets = {
    "all_without_alpha": [
        "sigma_noise",
        "A_repulsion",
        "gamma",
        "sigma_update",
        "eta_learning",
        "sigma_boundary"
    ],
    "all_with_alpha": [
        "sigma_noise",
        "A_repulsion",
        "gamma",
        "sigma_update",
        "eta_learning",
        "sigma_boundary",
        "alpha"
    ],
}

# =====================================================
# Helpers
# =====================================================
def find_result_file(folder, species, subject, sampler):
    search_dir = os.path.join(
        root_dir,
        folder,
        species,
        "results",
        sampler,
        subject
    )

    patterns = [
        os.path.join(search_dir, f"*{subject}*{sampler}*.pkl"),
        os.path.join(search_dir, "*.pkl"),
    ]

    files = []
    for pattern in patterns:
        files.extend(glob.glob(pattern))

    files = sorted(set(files))

    if len(files) == 0:
        return None

    return files[0]


def extract_params(data):
    if not isinstance(data, dict):
        return None

    if "best_params" in data:
        return data["best_params"]

    if "best_params_per_fold" in data:
        fold_params = data["best_params_per_fold"]

        if isinstance(fold_params, list) and len(fold_params) > 0:
            keys = fold_params[0].keys()

            return {
                k: np.nanmean([
                    fp[k] for fp in fold_params
                    if isinstance(fp, dict) and k in fp
                ])
                for k in keys
            }

    return None


def load_params(folder, species, subject, sampler):
    path = find_result_file(folder, species, subject, sampler)

    if path is None:
        return None, None

    with open(path, "rb") as f:
        data = pickle.load(f)

    params = extract_params(data)

    return params, path


def get_param(params, key, default=np.nan):
    if params is None:
        return default
    return float(params.get(key, default))


def safe_cosine(a, b):
    a = np.asarray(a, dtype=float)
    b = np.asarray(b, dtype=float)

    valid = np.isfinite(a) & np.isfinite(b)
    a = a[valid]
    b = b[valid]

    if len(a) == 0:
        return np.nan

    if np.allclose(a, 0) or np.allclose(b, 0):
        return np.nan

    return 1 - cosine(a, b)


def normalize_vector(vec, params):
    vec = np.asarray(vec, dtype=float).copy()

    for i, p in enumerate(params):
        if np.isfinite(vec[i]):
            low, high = param_ranges[p]
            vec[i] = (vec[i] - low) / (high - low)

    return vec


# =====================================================
# Load parameters
# =====================================================
rows = []

for species in species_list:
    for subject in participants_by_species[species]:
        for sampler in samplers:

            be_params, be_path = load_params(
                folders["BE"], species, subject, sampler
            )

            sc_params, sc_path = load_params(
                folders["SC"], species, subject, sampler
            )

            hybrid_params, hybrid_path = load_params(
                folders["Hybrid"], species, subject, sampler
            )

            if be_params is None or sc_params is None or hybrid_params is None:
                print("Missing:", species, subject, sampler)
                continue

            rows.append({
                "species": species,
                "subject": subject,
                "sampler": sampler,

                "BE_sigma_noise": get_param(be_params, "sigma_noise"),
                "BE_A_repulsion": get_param(be_params, "A_repulsion"),
                "BE_gamma": get_param(be_params, "gamma"),
                "BE_sigma_update": get_param(be_params, "sigma_update"),
                "BE_eta_learning": get_param(be_params, "eta_learning"),
                "BE_sigma_boundary": get_param(be_params, "sigma_boundary"),
                "BE_alpha": 0.0,

                "SC_sigma_noise": get_param(sc_params, "sigma_noise"),
                "SC_A_repulsion": get_param(sc_params, "A_repulsion"),
                "SC_gamma": get_param(sc_params, "gamma"),
                "SC_sigma_update": get_param(sc_params, "sigma_update"),
                "SC_eta_learning": get_param(sc_params, "eta_learning"),
                "SC_sigma_boundary": get_param(sc_params, "sigma_boundary"),
                "SC_alpha": 1.0,

                "Hybrid_sigma_noise": get_param(hybrid_params, "sigma_noise"),
                "Hybrid_A_repulsion": get_param(hybrid_params, "A_repulsion"),
                "Hybrid_gamma": get_param(hybrid_params, "gamma"),
                "Hybrid_sigma_update": get_param(hybrid_params, "sigma_update"),
                "Hybrid_eta_learning": get_param(hybrid_params, "eta_learning"),
                "Hybrid_sigma_boundary": get_param(hybrid_params, "sigma_boundary"),
                "Hybrid_alpha": get_param(hybrid_params, "alpha"),

                "BE_path": be_path,
                "SC_path": sc_path,
                "Hybrid_path": hybrid_path,
            })

df = pd.DataFrame(rows)

print("Loaded rows:", len(df))
display(df.head())

if len(df) == 0:
    raise ValueError("No files loaded. Check root_dir and folder structure.")

df.to_csv(
    os.path.join(save_dir, "loaded_parameter_table.csv"),
    index=False
)

# =====================================================
# Compute range-normalized cosine similarities
# =====================================================
cosine_rows = []

for _, row in df.iterrows():

    for vector_name, params in parameter_sets.items():

        BE_vec = np.array([row[f"BE_{p}"] for p in params], dtype=float)
        SC_vec = np.array([row[f"SC_{p}"] for p in params], dtype=float)
        H_vec = np.array([row[f"Hybrid_{p}"] for p in params], dtype=float)

        BE_norm = normalize_vector(BE_vec, params)
        SC_norm = normalize_vector(SC_vec, params)
        H_norm = normalize_vector(H_vec, params)

        cos_H_BE = safe_cosine(H_norm, BE_norm)
        cos_H_SC = safe_cosine(H_norm, SC_norm)
        cos_BE_SC = safe_cosine(BE_norm, SC_norm)

        cosine_rows.append({
            "species": row["species"],
            "subject": row["subject"],
            "sampler": row["sampler"],
            "vector_name": vector_name,
            "normalization": "range",
            "Hybrid_alpha": row["Hybrid_alpha"],

            "cos_Hybrid_BE": cos_H_BE,
            "cos_Hybrid_SC": cos_H_SC,
            "cos_BE_SC": cos_BE_SC,
            "delta_cos_SC_minus_BE": cos_H_SC - cos_H_BE,
        })

cos_df = pd.DataFrame(cosine_rows)

cos_df.to_csv(
    os.path.join(save_dir, "cosine_similarity_results_range_normalized.csv"),
    index=False
)

display(cos_df.head())

# =====================================================
# Summaries
# =====================================================
summary = (
    cos_df
    .groupby(["species", "vector_name"])
    [
        [
            "cos_Hybrid_BE",
            "cos_Hybrid_SC",
            "cos_BE_SC",
            "delta_cos_SC_minus_BE",
            "Hybrid_alpha"
        ]
    ]
    .agg(["mean", "std", "median"])
)

summary.to_csv(
    os.path.join(save_dir, "cosine_similarity_summary_range_normalized.csv")
)

display(summary)

# =====================================================
# Rank-sum tests: Human vs Mouse for delta cosine
# =====================================================
test_rows = []

for vector_name in cos_df["vector_name"].unique():

    sub = cos_df[cos_df["vector_name"] == vector_name]

    mouse_vals = sub[sub["species"] == "Mouse"]["delta_cos_SC_minus_BE"].dropna()
    human_vals = sub[sub["species"] == "Human"]["delta_cos_SC_minus_BE"].dropna()

    if len(mouse_vals) > 0 and len(human_vals) > 0:
        stat, p = ranksums(mouse_vals, human_vals)
    else:
        stat, p = np.nan, np.nan

    test_rows.append({
        "comparison": "Human vs Mouse",
        "vector_name": vector_name,
        "normalization": "range",
        "metric": "delta_cos_SC_minus_BE",
        "rank_sum_statistic": stat,
        "p_value": p,
        "n_mouse": len(mouse_vals),
        "n_human": len(human_vals),
    })

test_df = pd.DataFrame(test_rows)

test_df.to_csv(
    os.path.join(save_dir, "ranksum_human_vs_mouse_delta_cosine_range_normalized.csv"),
    index=False
)

display(test_df)

# =====================================================
# Spearman: alpha vs delta cosine
# =====================================================
corr_rows = []

for species in species_list:
    for vector_name in cos_df["vector_name"].unique():

        sub = cos_df[
            (cos_df["species"] == species) &
            (cos_df["vector_name"] == vector_name)
        ]

        valid = sub[
            ["Hybrid_alpha", "delta_cos_SC_minus_BE"]
        ].replace([np.inf, -np.inf], np.nan).dropna()

        if len(valid) >= 3:
            rho, p = spearmanr(
                valid["Hybrid_alpha"],
                valid["delta_cos_SC_minus_BE"]
            )
        else:
            rho, p = np.nan, np.nan

        corr_rows.append({
            "species": species,
            "vector_name": vector_name,
            "normalization": "range",
            "x": "Hybrid_alpha",
            "y": "delta_cos_SC_minus_BE",
            "spearman_rho": rho,
            "p_value": p,
            "n": len(valid),
        })

corr_df = pd.DataFrame(corr_rows)

corr_df.to_csv(
    os.path.join(save_dir, "alpha_vs_delta_cosine_correlations_range_normalized.csv"),
    index=False
)

display(corr_df)

# =====================================================
# Plot helpers
# =====================================================
def plot_delta_vs_alpha(vector_name):
    sub = cos_df[cos_df["vector_name"] == vector_name]

    plt.figure(figsize=(8, 6), dpi=300)

    for species, marker in [("Mouse", "o"), ("Human", "^")]:
        sp = sub[sub["species"] == species]

        plt.scatter(
            sp["Hybrid_alpha"],
            sp["delta_cos_SC_minus_BE"],
            marker=marker,
            s=80,
            alpha=0.8,
            label=species
        )

    plt.axhline(0, linestyle="--", linewidth=1)
    plt.axvline(0.5, linestyle="--", linewidth=1)

    plt.xlabel("Hybrid alpha")
    plt.ylabel("cos(Hybrid, SC) - cos(Hybrid, BE)")
    plt.title(f"Delta cosine vs alpha\n{vector_name}, range-normalized")
    plt.grid(alpha=0.3)
    plt.legend()
    plt.tight_layout()

    fname = f"delta_cos_vs_alpha_{vector_name}_range_normalized.png"
    path = os.path.join(save_dir, fname)

    plt.savefig(path, dpi=300)
    plt.show()

    print("Saved:", path)


def plot_cosine_pairs(vector_name):
    sub = cos_df[cos_df["vector_name"] == vector_name]

    plt.figure(figsize=(8, 6), dpi=300)

    metrics = [
        "cos_Hybrid_BE",
        "cos_Hybrid_SC",
        "cos_BE_SC"
    ]

    xlabels = [
        "Hybrid-BE",
        "Hybrid-SC",
        "BE-SC"
    ]

    for i, metric in enumerate(metrics):

        for species, marker, offset in [
            ("Mouse", "o", -0.08),
            ("Human", "^", 0.08)
        ]:
            vals = sub[sub["species"] == species][metric].dropna().to_numpy()
            jitter = np.random.normal(0, 0.025, len(vals))

            plt.scatter(
                np.full(len(vals), i + offset) + jitter,
                vals,
                marker=marker,
                s=60,
                alpha=0.75,
                label=species if i == 0 else None
            )

            if len(vals) > 1:
                plt.errorbar(
                    i + offset,
                    np.mean(vals),
                    yerr=np.std(vals, ddof=1),
                    fmt="k_",
                    capsize=6,
                    linewidth=2
                )
            elif len(vals) == 1:
                plt.errorbar(
                    i + offset,
                    np.mean(vals),
                    yerr=0,
                    fmt="k_",
                    capsize=6,
                    linewidth=2
                )

    plt.xticks(range(len(metrics)), xlabels)
    plt.ylabel("Cosine similarity")
    plt.title(f"Cosine similarity\n{vector_name}, range-normalized")
    plt.axhline(0, linestyle="--", linewidth=1)
    plt.grid(alpha=0.3)
    plt.legend()
    plt.tight_layout()

    fname = f"cosine_pairs_{vector_name}_range_normalized.png"
    path = os.path.join(save_dir, fname)

    plt.savefig(path, dpi=300)
    plt.show()

    print("Saved:", path)


# =====================================================
# Make plots
# =====================================================
for vector_name in [
    "all_without_alpha",
    "all_with_alpha",
]:
    plot_delta_vs_alpha(vector_name)
    plot_cosine_pairs(vector_name)

print("Done. Results saved in:")
print(save_dir)
```
