```python
import pickle
import pandas as pd

with open('SC_Data_Cleaned.pkl', 'rb') as file:
    data_mouse = pickle.load(file)


with open('Data_human_first_dists_V2.pkl', 'rb') as file:
    data_human = pickle.load(file)


with open('Data_rat_first_dists_V2.pkl', 'rb') as file:
    data_rat = pickle.load(file)


df_human = pd.DataFrame(data_human)
df_mouse = pd.DataFrame(data_mouse)
df_rat = pd.DataFrame(data_rat)

```

```python
human_lengths = (
    df_human.groupby("Participant_ID")
    .size()
    .reset_index(name="n_trials")
)

mouse_lengths = (
    df_mouse.groupby("Participant_ID")
    .size()
    .reset_index(name="n_trials")
)

rat_lengths = (
    df_rat.groupby("Participant_ID")
    .size()
    .reset_index(name="n_trials")
)

print(human_lengths)
print(mouse_lengths)
print(rat_lengths)
```
